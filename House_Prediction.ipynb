```python
# =====================================================
# STEP 4 : Read Dataset
# =====================================================

df = pd.read_csv("House Price Prediction Dataset.csv")
```

```python
# =====================================================
# STEP 5 : Display Dataset
# =====================================================

df.head()
```

```python
# =====================================================
# STEP 5 : Display Dataset
# =====================================================

df.head()
```

```python
# =====================================================
# STEP 7 : Check Missing Values
# =====================================================

df.isnull().sum()
```

```python
# =====================================================
# STEP 8 : Remove ID Column
# =====================================================

df = df.drop("Id", axis=1)
```

```python
# =====================================================
# STEP 9 : Separate Features and Target
# =====================================================

X = df.drop("Price", axis=1)
y = df["Price"]
```

```python
# =====================================================
# STEP 10 : Define Numerical & Categorical Columns
# =====================================================

categorical_features = [
    "Location",
    "Condition",
    "Garage"
]

numerical_features = [
    "Area",
    "Bedrooms",
    "Bathrooms",
    "Floors",
    "YearBuilt"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# =====================================================
# STEP 11 : Data Preprocessing
# =====================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ],
    remainder="passthrough"
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

# =====================================================
# STEP 12 : Create Machine Learning Pipeline
# =====================================================

model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ))
])
```

```python
from sklearn.model_selection import train_test_split

# =====================================================
# STEP 13 : Split Dataset
# =====================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# =====================================================
# STEP 14 : Train Model
# =====================================================

model.fit(X_train, y_train)
```

```python
# =====================================================
# STEP 15 : Make Predictions
# =====================================================

predictions = model.predict(X_test)
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# =====================================================
# STEP 16 : Model Evaluation
# =====================================================

mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, predictions)

print("Model Performance")
print("-"*30)

print("Mean Absolute Error :", mae)
print("Mean Squared Error :", mse)
print("Root Mean Squared Error :", rmse)
print("R2 Score :", r2)
```

```python
# =====================================================
# STEP 17 : Test Prediction
# =====================================================

sample = pd.DataFrame({
    "Area":[2500],
    "Bedrooms":[4],
    "Bathrooms":[3],
    "Floors":[2],
    "YearBuilt":[2018],
    "Location":["Urban"],
    "Condition":["Excellent"],
    "Garage":["Yes"]
})

prediction = model.predict(sample)

print("Predicted House Price : ₹{:,.2f}".format(prediction[0]))
```
